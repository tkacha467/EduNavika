When students learn a topic, their memory retention decays over time following an exponential curve:
$$R(t) = e^{-t / S}$$
where $t$ is elapsed time, $S$ is memory stability, and $R$ is retrieval probability.


### Step 1: Simulate the Ebbinghaus Forgetting Curve
Let's compute retention over 14 days for different stability levels.


```python
import math

def retention(t_days, stability):
    return math.exp(-t_days / stability)

stabilities = {'Novice (S=2)': 2.0, 'Moderate (S=5)': 5.0, 'Mastered (S=12)': 12.0}

print(f'{"Days":<6} | ' + ' | '.join([f'{k:<18}' for k in stabilities]))
print('-' * 65)
for day in [0, 1, 2, 4, 7, 10, 14]:
    row = [f'{day:<6}']
    for s in stabilities.values():
        r = retention(day, s) * 100
        row.append(f'{r:>6.1f}%' + ' ' * 11)
    print(' | '.join(row))
```

### Step 2: Spaced Repetition Trigger (< 80% Retention)
EduNavika triggers a review recommendation whenever predicted retention drops below **80%**.


```python
threshold = 0.80
for name, stability in stabilities.items():
    # R = exp(-t / S) => t = -S * ln(R)
    days_until_review = -stability * math.log(threshold)
    print(f'{name:<20}: Review recommended after {days_until_review:.1f} days')
```
