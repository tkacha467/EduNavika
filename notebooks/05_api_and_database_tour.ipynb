# 🗄️ Notebook 5: SQLite Database & Curriculum Models Tour
**EduNavika** -- Inspecting Curriculum Tables, Questions, and Attempt Tracking.

This notebook interacts directly with EduNavika's local SQLite database using SQLAlchemy ORM.


### Step 1: Connect to the EduNavika Database
Let's query the curriculum hierarchy (Standards, Subjects, Chapters, Topics).


```python
from backend.app.core.database import SessionLocal
from backend.app.models.curriculum_models import Standard, Subject, Chapter, Topic

db = SessionLocal()
standards = db.query(Standard).all()
print(f'Total Standards Registered: {len(standards)}')
for s in standards:
    print(f' - Standard {s.grade_number}: {s.name}')
```

### Step 2: Query Subjects & Topics for Standard 10
Let's see what topics exist under Standard 10 Mathematics and Science.


```python
std_10 = db.query(Standard).filter(Standard.grade_number == 10).first()
if std_10:
    print(f'Subjects under Standard 10:')
    for subj in std_10.subjects[:3]:
        print(f'  Subject: {subj.name} ({subj.code})')
        for chap in subj.chapters[:2]:
            print(f'    Chapter {chap.chapter_number}: {chap.title}')
            for top in chap.topics[:2]:
                print(f'      Topic: {top.title}')
else:
    print('Standard 10 record not yet in local DB. Ingest via pipeline first.')
```

### Step 3: Query Practice Questions and Assessments
Let's inspect MCQ questions stored in the database.


```python
from backend.app.models.assessment_models import MCQQuestion

questions = db.query(MCQQuestion).limit(3).all()
print(f'Total Questions Sampled: {len(questions)}')
for i, q in enumerate(questions, 1):
    print(f'\n[Q{i}] {q.question_text}')
    print(f' A) {q.option_a}')
    print(f' B) {q.option_b}')
    print(f' Correct: Option {q.correct_option.value}')

db.close()
```
